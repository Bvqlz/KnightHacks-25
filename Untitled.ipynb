```python
!pip install geopandas numpy ortools 
!pip install --pre matplotlib --upgrade
import geopandas as gpd
import numpy as np
import pandas as pd
from shapely import wkt
from ortools.constraint_solver import routing_enums_pb2
from ortools.constraint_solver import pywrapcp
import pickle
from pathlib import Path
import shutil

#| `distance_matrix.npy` | NumPy array (N x N) | Symmetric matrix of door-to-door flight distances between waypoints computed inside the polygon. Use this as the cost matrix in your optimizer. |
#| `predecessors.npy` | NumPy array (N x N) | For each pair of nodes `(i, j)`, the predecessor index on the shortest path from `i` to `j`. Use `scipy.sparse.csgraph.dijkstra` conventions to expand indirect legs back into actual waypoint sequences. |
#| `points_lat_long.npy` | NumPy array (N x 2) | Geographic coordinates (longitude, latitude) for every waypoint index. Index into this after planning to obtain real-world coordinates. |
#| `asset_indexes.npy` | NumPy array | Subset of waypoint indices corresponding to electrical assets (i.e. poles), formatted as a slice: `[first index, last index]`. |
#| `photo_indexes.npy` | NumPy array | Subset of waypoint indices for the 4 photo points around each asset, formatted as a slice: `[first index, last index]`.  |
#| `polygon_lon_lat.wkt` | Text (WKT) | WKT polygon describing the allowed flight region. Use Shapely/GeoPandas to visualize or enforce constraints. |



# --- Load data ---
print("=== Loading Data ===")
distance_matrix = np.load('distance_matrix.npy')
asset_indexes = np.load('asset_indexes.npy')
photo_indexes = np.load('photo_indexes.npy')
points_lat_long = np.load('points_lat_long.npy')
predecessors = np.load('predecessors.npy')
waypoint_indexes = np.load('waypoint_indexes.npy')

print(f"distance_matrix: {distance_matrix.shape}")
print(f"asset_indexes: {asset_indexes}")
print(f"photo_indexes: {photo_indexes}")
print(f"points_lat_long: {points_lat_long.shape}")
print(f"predecessors: {predecessors.shape}")
print(f"waypoint_indexes: {waypoint_indexes}")

# --- Understand the data structure ---
num_navigable = distance_matrix.shape[0]  # 2779 waypoints we can visit
num_total_points = points_lat_long.shape[0]  # 4274 total points
num_assets = asset_indexes[1] - asset_indexes[0] + 1  # Assets (not navigable)
num_photos = photo_indexes[1] - photo_indexes[0] + 1  # Photo waypoints (navigable)

print(f"\n=== Data Structure ===")
print(f"Total points in coordinate list: {num_total_points}")
print(f"Navigable waypoints (in distance matrix): {num_navigable}")
print(f"Photo waypoints: {num_photos} (indices {photo_indexes[0]} to {photo_indexes[1]})")
print(f"Asset points: {num_assets} (indices {asset_indexes[0]} to {asset_indexes[1]})")
print(f"Note: Assets are NOT in distance matrix - they're reference points only")

# --- Visualize the problem ---
try:
    # Separate navigable waypoints from asset locations
    navigable_coords = points_lat_long[:num_navigable]
    asset_coords = points_lat_long[asset_indexes[0]:asset_indexes[1]+1]
    
    fig, ax = plt.subplots(figsize=(12, 10))
    
    # Load polygon boundary
    try:
        with open("polygon_lon_lat.wkt") as f:
            polygon = wkt.loads(f.read())
        gdf_poly = gpd.GeoDataFrame(geometry=[polygon], crs="EPSG:4326")
        gdf_poly.plot(ax=ax, edgecolor='blue', facecolor='none', linewidth=2)
    except FileNotFoundError:
        print("Polygon file not found")
    
    # Plot navigable waypoints (red) and assets (green)
    ax.scatter(navigable_coords[:, 0], navigable_coords[:, 1], 
               c='red', s=10, alpha=0.5, label=f'Navigable waypoints ({num_navigable})')
    ax.scatter(asset_coords[:, 0], asset_coords[:, 1], 
               c='green', s=50, marker='s', alpha=0.7, label=f'Assets ({num_assets})')
    
    ax.set_xlabel('Longitude')
    ax.set_ylabel('Latitude')
    ax.set_title('Drone Inspection Area')
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
except Exception as e:
    print(f"Visualization error: {e}")

# --- Caching utilities ---
def cache_save(key, result, cache_dir="cache"):
    """Save result to cache with atomic write to prevent corruption."""
    Path(cache_dir).mkdir(exist_ok=True)
    cache_file = Path(cache_dir) / f"{key}.pkl"
    temp_file = Path(cache_dir) / f"{key}.pkl.tmp"
    
    try:
        # Write to temporary file first
        with open(temp_file, "wb") as f:
            pickle.dump(result, f)
        
        # Atomic rename (safer than direct write)
        shutil.move(str(temp_file), str(cache_file))
        print(f" Solution cached to: {cache_file}")
    except Exception as e:
        print(f"  Cache save failed: {e}")
        if temp_file.exists():
            temp_file.unlink()


def cache_load(key, cache_dir="cache"):
    """Load cached result with error handling."""
    cache_file = Path(cache_dir) / f"{key}.pkl"
    
    if not cache_file.exists():
        return None
    
    try:
        with open(cache_file, "rb") as f:
            result = pickle.load(f)
        print(f" Loaded cached solution from: {cache_file}")
        return result
    except (EOFError, pickle.UnpicklingError, Exception) as e:
        print(f"  Corrupted cache detected ({e}). Deleting and re-solving...")
        try:
            cache_file.unlink()
        except:
            pass
        return None


def clear_cache(cache_dir="cache"):
    """Delete all cached solutions."""
    cache_path = Path(cache_dir)
    if cache_path.exists():
        shutil.rmtree(cache_path)
        print(f"  Cache cleared: {cache_dir}")


# --- Create data model ---
def create_data_model(subset_size=None):
    """Stores the data for the problem."""
    data = {}
    
    # Option to solve smaller subset for testing
    if subset_size and subset_size < num_navigable:
        print(f"\n  Using subset of {subset_size} waypoints for faster solving")
        data["distance_matrix"] = distance_matrix[:subset_size, :subset_size].astype(np.int32)
        data["num_waypoints"] = subset_size
    else:
        data["distance_matrix"] = distance_matrix.astype(np.int32)
        data["num_waypoints"] = num_navigable
    
    data["num_vehicles"] = 1
    data["depot"] = 0  # Starting waypoint
    
    # Calculate reasonable max distance
    total_dist = np.sum(distance_matrix)
    avg_dist = np.mean(distance_matrix[distance_matrix > 0])
    max_dist = int(avg_dist * data["num_waypoints"] * 1.5)
    
    data["max_distance"] = max_dist
    
    print(f"\nDistance matrix stats:")
    print(f"  Min distance: {np.min(distance_matrix[distance_matrix > 0]):.1f}m")
    print(f"  Max distance: {np.max(distance_matrix):.1f}m")
    print(f"  Avg distance: {avg_dist:.1f}m")
    print(f"  Set max route: {max_dist:.1f}m")
    
    return data


def print_solution(data, manager, routing, solution, points_lat_long):
    """Prints solution details."""
    print("\n" + "="*50)
    print("=== SOLUTION FOUND ===")
    print("="*50)
    print(f"Objective value (total distance): {solution.ObjectiveValue():,.0f}m")
    print(f"                                   {solution.ObjectiveValue()/1000:.2f}km")
    
    route = []
    index = routing.Start(0)
    route_distance = 0
    
    print(f"\nRoute visiting {data['num_waypoints']} waypoints:")
    print("(Showing first 10 and last 10 waypoints...)\n")
    
    waypoint_count = 0
    while not routing.IsEnd(index):
        node = manager.IndexToNode(index)
        route.append(node)
        
        if waypoint_count < 10 or waypoint_count >= data['num_waypoints'] - 10:
            coord = points_lat_long[node]
            print(f"  {waypoint_count:4d}. Waypoint {node:4d}: [{coord[0]:.6f}, {coord[1]:.6f}]")
        elif waypoint_count == 10:
            print(f"  ... ({data['num_waypoints'] - 20} waypoints omitted) ...")
        
        previous_index = index
        index = solution.Value(routing.NextVar(index))
        route_distance += routing.GetArcCostForVehicle(previous_index, index, 0)
        waypoint_count += 1
    
    # Add final node (return to depot)
    final_node = manager.IndexToNode(index)
    route.append(final_node)
    coord = points_lat_long[final_node]
    print(f"  {waypoint_count:4d}. Waypoint {final_node:4d}: [{coord[0]:.6f}, {coord[1]:.6f}] (END)")
    
    print(f"\n{'='*50}")
    print(f"Total route distance: {route_distance:,.0f}m ({route_distance/1000:.2f}km)")
    print(f"Total waypoints visited: {len(route)}")
    print(f"{'='*50}\n")
    
    return route


def visualize_route(route, points_lat_long, sample_rate=10):
    !pip install matplotlib
    import matplotlib.pyplot as plt
    import numpy as np

    """Visualize the optimized route (sampling for large routes)."""
    route_coords = points_lat_long[route]
    
    # Sample route for visualization if too large
    if len(route) > 500:
        sampled_indices = list(range(0, len(route), sample_rate))
        if sampled_indices[-1] != len(route) - 1:
            sampled_indices.append(len(route) - 1)
        route_coords_plot = route_coords[sampled_indices]
        print(f"Visualizing {len(sampled_indices)} of {len(route)} waypoints")
    else:
        route_coords_plot = route_coords
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))
    
    # Left plot: Full context
    ax1.scatter(points_lat_long[:num_navigable, 0], points_lat_long[:num_navigable, 1], 
                c='lightgray', s=5, alpha=0.3, label='All waypoints')
    ax1.plot(route_coords_plot[:, 0], route_coords_plot[:, 1], 
             'b-', linewidth=1.5, alpha=0.7, label='Route')
    ax1.scatter(route_coords[0, 0], route_coords[0, 1], 
                c='green', s=200, marker='*', label='Start', zorder=5)
    ax1.scatter(route_coords[-1, 0], route_coords[-1, 1], 
                c='red', s=200, marker='X', label='End', zorder=5)
    ax1.set_xlabel('Longitude')
    ax1.set_ylabel('Latitude')
    ax1.set_title('Complete Route Overview')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # Right plot: Route detail (first 100 waypoints)
    detail_size = min(100, len(route_coords))
    ax2.plot(route_coords[:detail_size, 0], route_coords[:detail_size, 1], 
             'b-o', linewidth=1, markersize=3, alpha=0.7)
    ax2.scatter(route_coords[0, 0], route_coords[0, 1], 
                c='green', s=200, marker='*', label='Start', zorder=5)
    ax2.set_xlabel('Longitude')
    ax2.set_ylabel('Latitude')
    ax2.set_title(f'Route Detail (First {detail_size} Waypoints)')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()


def findSolution(routing, search_parameters, key="tsp_solution"):
    """Find or load a cached OR-Tools routing solution."""
    # Try to load cached solution first
    cached = cache_load(key)
    
    if cached is not None:
        return cached
    
    # No cache found — solve fresh
    print(" No cache found. Solving fresh...")
    solution = routing.SolveWithParameters(search_parameters)
    
    # Save for next time (only if solution found)
    if solution:
        cache_save(key, solution)
    else:
        print(" No solution found — nothing cached.")
    
    return solution


def main(use_subset=False, subset_size=500):
    """Entry point of the program."""
    
    # For testing with large dataset, use a subset
    data = create_data_model(subset_size if use_subset else None)
    
    # Create cache key based on problem size
    cache_key = f"route_{data['num_waypoints']}_waypoints"
    
    print(f"\n{'='*50}")
    print("=== OPTIMIZATION PARAMETERS ===")
    print(f"{'='*50}")
    print(f"Number of waypoints to visit: {data['num_waypoints']}")
    print(f"Number of vehicles (drones): {data['num_vehicles']}")
    print(f"Maximum route distance: {data['max_distance']:,.0f}m ({data['max_distance']/1000:.1f}km)")
    print(f"Starting depot: Waypoint {data['depot']}")
    print(f"Cache key: {cache_key}")
    print(f"{'='*50}\n")
    
    # Create the routing index manager
    manager = pywrapcp.RoutingIndexManager(
        len(data["distance_matrix"]), 
        data["num_vehicles"], 
        data["depot"]
    )
    
    # Create Routing Model
    routing = pywrapcp.RoutingModel(manager)
    
    # Create and register a transit callback
    def distance_callback(from_index, to_index):
        """Returns the distance between the two nodes."""
        from_node = manager.IndexToNode(from_index)
        to_node = manager.IndexToNode(to_index)
        return data["distance_matrix"][from_node][to_node]
    
    transit_callback_index = routing.RegisterTransitCallback(distance_callback)
    
    # Define cost of each arc
    routing.SetArcCostEvaluatorOfAllVehicles(transit_callback_index)
    
    # Add Distance constraint
    dimension_name = "Distance"
    routing.AddDimension(
        transit_callback_index,
        0,  # no slack
        data["max_distance"],  # vehicle maximum travel distance
        True,  # start cumul to zero
        dimension_name,
    )
    distance_dimension = routing.GetDimensionOrDie(dimension_name)
    distance_dimension.SetGlobalSpanCostCoefficient(100)
    
    # Setting search parameters
    search_parameters = pywrapcp.DefaultRoutingSearchParameters()
    search_parameters.first_solution_strategy = (
        routing_enums_pb2.FirstSolutionStrategy.PATH_CHEAPEST_ARC
    )
    search_parameters.local_search_metaheuristic = (
        routing_enums_pb2.LocalSearchMetaheuristic.GUIDED_LOCAL_SEARCH
    )
    search_parameters.time_limit.seconds = 60  # Increase for larger problems
    search_parameters.log_search = True  # Show progress
    
    print(" Starting route optimization...")
    print("(This may take a while for 2779 waypoints...)\n")
    
    # Find solution (with caching)
    solution = findSolution(routing, search_parameters, key=cache_key)
    
    # Print and visualize solution
    if solution:
        route = print_solution(data, manager, routing, solution, points_lat_long)
        
        # Save route to file
        np.save('optimized_route.npy', np.array(route))
        print(f" Route saved to 'optimized_route.npy'")
        
        # Visualize
        visualize_route(route, points_lat_long)
        
        return route
    else:
        print(" No solution found!")
        print("\nTroubleshooting:")
        print("1. The max_distance constraint may be too restrictive")
        print("2. Try running with use_subset=True first")
        print("3. Increase time_limit in search_parameters")
        return None


if __name__ == "__main__":
    # For initial testing with 2779 waypoints, start with a subset
    print("Choose mode:")
    print("1. Test with 500 waypoints (fast, ~1 min)")
    print("2. Full optimization with all 2779 waypoints (slow, ~5-10 min)")
    print("3. Clear cache and re-solve")
    
    # Change this to False to run full optimization
    TEST_MODE = False
    CLEAR_CACHE = False  # Set to True to delete corrupted cache
    
    if CLEAR_CACHE:
        clear_cache()
    
    if TEST_MODE:
        print("\n Running in TEST MODE (500 waypoints)\n")
        route = main(use_subset=True, subset_size=500)
    else:
        print("\n Running FULL OPTIMIZATION (2779 waypoints)\n")
        route = main(use_subset=False)
```
