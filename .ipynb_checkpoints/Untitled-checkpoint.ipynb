```python
# Install dependencies (only needed once per environment)
!pip install geopandas numpy geodatasets pandas matplotlib

# Df1- `distance_matrix.npy` | NumPy array (N x N) | Symmetric matrix of door-to-door flight distances between waypoints computed inside the polygon. Use this as the cost matrix in your optimizer. |
# Df2- `predecessors.npy` | NumPy array (N x N) | For each pair of nodes `(i, j)`, stores the predecessor index on the shortest path from `i` to `j`. Use `scipy.sparse.csgraph.dijkstra` conventions to expand indirect legs back into actual waypoint sequences. |
# Df3- `points_lat_long.npy` | NumPy array (N x 2) | Geographic coordinates (longitude, latitude) for every waypoint index. Index into this after planning to obtain real-world coordinates. |
# Df4- `asset_indexes.npy` | NumPy array | Subset of waypoint indices corresponding to electrical assets (i.e. poles), formatted as a slice: `[first index, last index]`. |
# Df5- `photo_indexes.npy` | NumPy array | Subset of waypoint indices for the 4 photo points around each asset, formatted as a slice: `[first index, last index]`.  |
# Df6- `polygon_lon_lat.wkt` | Text (WKT) | WKT polygon describing the allowed flight region. Use Shapely/GeoPandas to visualize or enforce constraints. |



import geopandas as gpd
import numpy as np
import pandas as pd
from shapely import wkt
import matplotlib.pyplot as plt

# --- Load data ---
distance_matrix = np.load('distance_matrix.npy')
asset_indexes = np.load('asset_indexes.npy')
photo_indexes = np.load('photo_indexes.npy')
points_lat_long = np.load('points_lat_long.npy')
predecessors = np.load('predecessors.npy')
waypoint_indexes = np.load('waypoint_indexes.npy')

# Print shapes to verify structure
print("distance_matrix:", distance_matrix.shape)
print("asset_indexes:", asset_indexes.shape)
print("photo_indexes:", photo_indexes.shape)
print("points_lat_long:", points_lat_long.shape)
print("predecessors:", predecessors.shape)
print("waypoint_indexes:", waypoint_indexes.shape)

# --- Convert all numpy arrays into pandas DataFrames ---
arrays = [distance_matrix, asset_indexes, photo_indexes, points_lat_long, predecessors, waypoint_indexes]
data_dict = {}

for i, arr in enumerate(arrays, start=1):
    df_name = f'df{i}'
    cols = [f'col_{j}' for j in range(arr.shape[1])] if arr.ndim > 1 else ['value']
    data_dict[df_name] = pd.DataFrame(arr, columns=cols)

# --- Convert the waypoint coordinates into a GeoDataFrame ---
gdf_points = gpd.GeoDataFrame(

    #These are the red dots. They represent points_lat_long, which represent where the drone has to go. 

    data_dict['df4'],  # df4 = points_lat_long
    geometry=gpd.points_from_xy(
        data_dict['df4']['col_0'],  # longitude
        data_dict['df4']['col_1']   # latitude
    ),
    crs="EPSG:4326"
)


# --- Identify asset and photo points safely ---
# Handles both slice ([start, end]) and explicit list formats

# Assets
if len(asset_indexes) == 2:
    asset_range = np.arange(asset_indexes[0], asset_indexes[1] + 1)
else:
    asset_range = np.array(asset_indexes)

# Photos
if len(photo_indexes) == 2:
    photo_range = np.arange(photo_indexes[0], photo_indexes[1] + 1)
else:
    photo_range = np.array(photo_indexes)

# --- Load the polygon boundary (optional) ---
try:
    with open("polygon_lon_lat.wkt") as f:
        polygon = wkt.loads(f.read())
    gdf_poly = gpd.GeoDataFrame(geometry=[polygon], crs="EPSG:4326")
    ax = gdf_poly.plot(edgecolor='blue', facecolor='none', figsize=(8, 8))
    # Blue is the flight boundary. 
except FileNotFoundError:
    print("Polygon file not found; plotting points only.")
    fig, ax = plt.subplots(figsize=(8, 8))

# --- Plot all point categories with color and legend ---
# Plot main waypoints
gdf_points.plot(ax=ax, color='gray', markersize=10, label='Waypoints')

# Plot assets (subset of gdf_points)
gdf_points.iloc[asset_range].plot(ax=ax, color='orange', markersize=30, label='Assets')

# Plot photo points (subset of gdf_points)
gdf_points.iloc[photo_range].plot(ax=ax, color='red', markersize=20, label='Photo Points')

# --- Add legend and labels ---
plt.legend()
plt.title("Drone Waypoints, Assets, and Photo Points")
plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.show()
```
